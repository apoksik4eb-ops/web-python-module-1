# ЗАДАНИЕ: Консольное приложение для учета складских заявок на выдачу материалов

Необходимо разработать консольное приложение WarehouseRequest Console для учета материалов на складе и обработки заявок на их выдачу сотрудникам организации.

В компании есть внутренний склад, на котором хранятся различные материалы: бумага, картриджи, кабели, канцелярские товары, расходники, комплектующие и другие позиции. Сотрудники разных отделов могут создавать заявки на получение материалов. Кладовщик должен видеть остатки на складе, обрабатывать заявки, пополнять материалы и списывать их после выдачи.

Приложение должно позволять вести учет материалов, создавать заявки на выдачу, контролировать остатки, изменять статусы заявок и получать краткую статистику.

## Цель работы:
Создать консольную программу, которая позволяет добавлять материалы на склад, просматривать остатки, создавать и обрабатывать заявки на выдачу материалов, а также сохранять данные между запусками программы.

## Функциональные требования:

### 1. Главное меню

При запуске программы пользователь должен видеть меню:

```txt
=== WarehouseRequest Console ===

1. Добавить материал на склад
2. Показать все материалы
3. Создать заявку на выдачу
4. Показать все заявки
5. Найти заявку
6. Изменить статус заявки
7. Пополнить остаток материала
8. Списать материал по заявке
9. Показать статистику
10. Сохранить данные
0. Выход
```

Пользователь выбирает действие путем ввода номера пункта меню.

### 2. Добавление материала на склад

Программа должна позволять добавить новый материал на склад.

При добавлении материала пользователь вводит следующие данные:

```txt
- название материала
- категория материала
- единица измерения
- количество на складе
- минимальный остаток
```

**Категория материала может быть любой, например:**

- канцелярия
- расходники
- техника
- комплектующие
- упаковка
- другое

**Единица измерения может быть любой, например:**

- шт.
- пачка
- коробка
- метр
- кг

**Каждому материалу должен автоматически присваиваться уникальный номер.**

**Пример материала:**

```txt
Номер материала: 1
Название: Бумага A4
Категория: канцелярия
Единица измерения: пачка
Количество на складе: 25
Минимальный остаток: 5
```

### 3. Просмотр всех материалов

Программа должна выводить список всех материалов, которые есть на складе.

Для каждого материала необходимо отображать:

```txt
- номер материала
- название материала
- категория
- единица измерения
- количество на складе
- минимальный остаток
```

Если материалов нет, программа должна вывести сообщение:

`Материалы отсутствуют.`

Если количество материала меньше или равно минимальному остатку, рядом с материалом желательно выводить предупреждение:

`Требуется пополнение`

**Пример вывода:**

```txt
Номер материала: 2
Название: Картридж HP 106A
Категория: расходники
Единица измерения: шт.
Количество на складе: 2
Минимальный остаток: 3
Статус остатка: Требуется пополнение
```

### 4. Создание заявки на выдачу

Программа должна позволять создать заявку на выдачу материала сотруднику.

При создании заявки пользователь вводит:

```txt
- ФИО сотрудника
- отдел сотрудника
- номер материала
- количество материала
- причина выдачи
- статус заявки
```

**Статус заявки может принимать одно из значений:**

- новая
- одобрена
- отклонена
- выполнена

**Каждой заявке должен автоматически присваиваться уникальный номер.**

Перед созданием заявки программа должна проверить:

1. Существует ли материал с указанным номером.
2. Количество запрашиваемого материала должно быть больше 0.
3. Запрашиваемое количество не должно превышать количество материала на складе.

Если все проверки пройдены, заявка создается.

**Пример заявки:**

```txt
Номер заявки: 1
Сотрудник: Иванов Иван Иванович
Отдел: Бухгалтерия
Материал: Бумага A4
Количество: 3 пачка
Причина выдачи: Печать отчетных документов
Статус: новая
```

### 5. Просмотр всех заявок

Программа должна выводить список всех созданных заявок.

Для каждой заявки необходимо отображать:

```txt
- номер заявки
- ФИО сотрудника
- отдел
- название материала
- количество материала
- причина выдачи
- статус заявки
```

Если заявок нет, программа должна вывести сообщение:

`Заявки отсутствуют.`

### 6. Поиск заявки

Программа должна позволять искать заявки по разным параметрам.

Необходимо реализовать поиск:

```txt
- по ФИО сотрудника
- по отделу
- по названию материала
- по статусу заявки
```

При выборе пункта `Найти заявку` пользователь должен выбрать критерий поиска, затем ввести значение для поиска.

Пример меню поиска:

```txt
1. Поиск по сотруднику
2. Поиск по отделу
3. Поиск по материалу
4. Поиск по статусу
```

Если подходящие заявки найдены, программа выводит их список.

Если заявки не найдены, программа выводит сообщение:

`Заявки не найдены.`

### 7. Изменение статуса заявки

Программа должна позволять изменить статус существующей заявки.

Для изменения статуса пользователь вводит номер заявки, после чего выбирает новый статус:

```txt
1. новая
2. одобрена
3. отклонена
4. выполнена
```

Если заявка с указанным номером существует, программа изменяет ее статус и выводит сообщение:

`Статус заявки успешно изменен.`

Если заявка не найдена, программа выводит сообщение:

`Заявка с указанным номером не найдена.`

### 8. Пополнение остатка материала

Программа должна позволять увеличить количество материала на складе.

Для пополнения остатка пользователь вводит:

```txt
- номер материала
- количество для добавления
```

Если материал с указанным номером существует, программа увеличивает его остаток и выводит сообщение:

`Остаток материала успешно пополнен.`

Если материал не найден, программа выводит сообщение:

`Материал с указанным номером не найден.`

Если пользователь вводит отрицательное количество или ноль, программа должна вывести сообщение:

`Количество для пополнения должно быть больше нуля.`

### 9. Списание материала по заявке

Программа должна позволять списать материал со склада по заявке.

Для списания пользователь вводит номер заявки.

Списание возможно только если:

```txt
- заявка существует
- статус заявки: одобрена
- на складе достаточно материала
```

После успешного списания:

```txt
- количество материала на складе уменьшается
- статус заявки меняется на выполнена
```

Если заявка не найдена, программа выводит сообщение:

`Заявка с указанным номером не найдена.`

Если заявка не одобрена, программа выводит сообщение:

`Списание возможно только по одобренной заявке.`

Если материала на складе недостаточно, программа выводит сообщение:

`Недостаточно материала на складе.`

Если списание выполнено успешно, программа выводит сообщение:

`Материал успешно списан по заявке.`

### 10. Статистика

Программа должна выводить краткую статистику:

```txt
- общее количество материалов
- общее количество заявок
- количество новых заявок
- количество одобренных заявок
- количество отклоненных заявок
- количество выполненных заявок
- количество материалов, требующих пополнения
- общее количество единиц материалов на складе
```

Пример вывода:

```txt
=== Статистика ===

Всего материалов: 8
Всего заявок: 15
Новые заявки: 4
Одобренные заявки: 3
Отклоненные заявки: 2
Выполненные заявки: 6
Материалы, требующие пополнения: 2
Всего единиц материалов на складе: 147
```

### 11. Сохранение данных

Программа должна сохранять данные о материалах и заявках в файл.

Формат файла можно выбрать самостоятельно:

- txt
- csv
- json

После сохранения программа должна вывести сообщение:

`Данные успешно сохранены.`

### 12. Загрузка данных

При запуске программы необходимо загружать ранее сохраненные материалы и заявки из файла, если такой файл существует.

Если файл отсутствует, программа начинает работу с пустыми списками материалов и заявок.

### 13. Выход из программы

При выборе пункта `0. Выход` программа должна завершить работу.

Перед выходом желательно выполнить автоматическое сохранение данных в файл.

1. Программа не должна завершаться аварийно при некорректном вводе пользователя.

2. Если пользователь вводит неправильный пункт меню, необходимо вывести сообщение:

`Некорректный выбор. Попробуйте снова.`

3. Если пользователь вводит несуществующий номер материала, необходимо вывести сообщение:

`Заявка с указанным номером не найдена.`

5. Если пользователь вводит недопустимый статус заявки, программа должна попросить повторить ввод.

6. Если пользователь вводит отрицательное количество материала, необходимо вывести сообщение:

`Количество материала не может быть отрицательным.`

7. Если пользователь вводит количество заявки меньше или равное нулю, необходимо вывести сообщение:

`Количество в заявке должно быть больше нуля.`

8. Если запрашиваемое количество превышает остаток на складе, необходимо вывести сообщение:

`Запрашиваемое количество превышает остаток на складе.`

9. Если пользователь вводит отрицательный минимальный остаток, необходимо вывести сообщение:

`Минимальный остаток не может быть отрицательным.`

## Дополнительные требования:

1. Номер материала должен быть уникальным.
2. Номер заявки должен быть уникальным.
3. После удаления или изменения заявки номера остальных заявок не должны изменяться.
4. После добавления нового материала его номер должен быть больше, чем у всех уже существующих материалов.
5. После добавления новой заявки ее номер должен быть больше, чем у всех уже существующих заявок.
6. Поиск по текстовым полям должен работать без учета регистра букв.
7. Количество материала на складе не может быть отрицательным.
8. Запрашиваемое количество материала должно быть больше 0.
9. Списание материала должно уменьшать остаток на складе.
10. Выполненная заявка не должна списываться повторно.
11. Отклоненная заявка не должна списываться.
12. Данные должны сохраняться между запусками программы.

## Пример сценария работы программы:

1. Пользователь запускает программу.
2. Программа загружает материалы и заявки из файла.

15. Программа сохраняет данные в файл.

## Результат выполнения работы:

В результате должно быть создано консольное приложение WarehouseRequest Console, которое позволяет вести учет складских материалов, создавать заявки на их выдачу, контролировать остатки, списывать материалы по заявкам и сохранять данные между запусками программы.

````python
class Request:
    def __init__(self, id, employeeName, department, materialId, quantity, reason, status="новая"):
        self.id = id
        self.employeeName = employeeName
        self.department = department
        self.materialId = materialId
        self.quantity = quantity
        self.reason = reason
        self.status = status

    def __str__(self):
        return (
            f"номер заявки: {self.id}\n"
            f"ФИО сотрудника: {self.employeeName}\n"
            f"отдел: {self.department}\n"
            f"название материала: {self.materialId}\n"
            f"количество материала: {self.quantity}\n"
            f"причина выдачи: {self.reason}\n"
            f"статус заявки: {self.status}\n"
        )

class Material:
    def __init__(self, id, name, category, unit, quantity, minQuantity):
        self.id = id
        self.name = name
        self.category = category
        self.unit = unit
        self.quantity = quantity
        self.minQuantity = minQuantity

    def __str__(self):
        return (
            f"номер материала: {self.id}\n"
            f"название материала: {self.name}\n"
            f"категория: {self.category}\n"
            f"еденица измерения: {self.unit}\n"
            f"количество на складе: {self.quantity}\n"
            f"минимальный остаток: {self.minQuantity}\n"
        )
    
class MaterialService:
    def __init__(self):
        self.materials = []
        self.next_material_id = 1

    def add_material(self, name, category, unit, quantity, minQuantity):
        material = Material(self.next_material_id, name, category, unit, quantity, minQuantity)
        self.materials.append(material)
        print(f"Материал {self.next_material_id} добавлен.")
        self.next_material_id += 1

    def show_all_material(self):
        if not self.materials:
            print("Материалы отсутствуют.")
            return
        
        print("Все материалы:\n")
        for material in self.materials:
            print(material)

class RequestService:
    def __init__(self):
        self.requests = []
        self.next_request_id = 1

    def add_request(self, employeeName, department, materialId, quantity, reason):
        request = Request(self.next_request_id, employeeName, department, materialId, quantity, reason)
        self.requests.append(request)
        print(f"Заявка {self.next_request_id} создана.")
        self.next_request_id += 1
    
    def show_all_request(self):
        if not self.requests:
            print("Заявки отсутствуют.")
            return
        
        print("Все заявки:\n")
        for request in self.requests:
            print(request)


class ConsoleMenu:
    def __init__(self, material_service: MaterialService, request_service: RequestService):
        self.service_material = material_service
        self.service_request = request_service
        
    def add_material(self):
        name = input("Введите название материала: ")
        category = input("Введите категорию материала: ")
        unit = input("Введите еденицу измерения: ")
        quantity = input("Введите количество на складе: ")
        minQuantity = input("Введите минимальный остаток: ")

        self.service_material.add_material(name, category, unit, quantity, minQuantity)

    def find_request(self):
        choice = input("Выберите критерий поиска:\n1. по ФИО сотрудника\n2. по отделу\n3. по названию материала\n4. по статусу заявки\n")

        found_request = []
        if choice == "1":
            found = input("Введите ФИО сотрудника: ")
            for request in request_service.requests:
                if request.employeeName == found:
                    found_request.append(request)
        elif choice == "2":
            found = input("Введите название отдела: ")
            for request in request_service.requests:
                if request.department == found:
                    found_request.append(request)
        elif choice == "3":
            found = input("Введите название материала: ")
            for request in request_service.requests:
                if request.materialId == found:
                    found_request.append(request)
        elif choice == "4":
            found = input("Введите статус заявки")
            for request in request_service.requests:
                if request.status == found:
                    found_request.append(request)
        else:
            print("Неправильный выбор.")
            return
        
        if found_request:
            print("Найденные заявки по вашему запросу:")
            for request in found_request:
                print(request)
        else:
            print("Заявки не найдены.")

    def add_request(self):
        employeeName = input("Введите ФИО сотрудника: ")
        department = input("Введите отдел: ")
        materialId = input("Введите номер материала: ")
        quantity = input("Введите количество материала: ")
        reason = input("Введите причину выдачи: ")
        
        self.service_request.add_request(employeeName, department, materialId, quantity, reason)

    def change_request_status(self):
        change_request = int(input("Введите номер заявки: "))

        for request in request_service.requests:
            if request.id == change_request:
                choice = input(f"Выеберите новый статус для {change_request} заявки:\n1. новая\n2. одобрена\n3. отклонена\n4. выполнена\n ")
                if choice == "1":
                    request.status = "новая"
                elif choice == "2":
                    request.status = "одобрена"
                elif choice == "3":
                    request.status = "отклонена"
                elif choice == "4":
                    request.status = "выполнена"
                else:
                    print("Неверный выбор.")
                    return
                
                print("Статус заявки успешно изменен.")
                return
            
        print("Заявка c указанным номером не найдена.")  

    def statistic(self):
        total_materials = 0
        total_requests = 0
        new_requests = 0
        approved_requests = 0
        rejected_requests = 0
        complete_requests = 0
        requiring_add_materials = 0
        total_all_materials = 0

        for material in material_service.materials:
            total_materials += 1

        for request in request_service.requests:
            total_requests += 1
            if request.status == "новая":
                new_requests += 1
            elif request.status == "одобрена":
                approved_requests += 1
            elif request.status == "отклонена":
                rejected_requests += 1
            elif request.statsus == "выполнена":
                complete_requests += 1

    def show_menu(self):
        print("\n=== WarehouseRequest Console ===\n")
        print("1. Добавить материал на склад")
        print("2. Показать все материалы")
        print("3. Создать заявку на выдачу")
        print("4. Показать все заявки")
        print("5. Найти заявку")
        print("6. Изменить статус заявки")
        print("7. Пополнить остаток материала")
        print("8. Списать материал по заявке")
        print("9. Показать статистику")
        print("10. Сохранить данные")
        print("0. Выход")

    def main(self):
        while True:
            self.show_menu()

            choice = input("Выберите пункт меню: ")

            if choice == "1":
                self.add_material()
            elif choice == "2":
                self.service_material.show_all_material()
            elif choice == "3":
                self.add_request()
            elif choice == "4":
                self.service_request.show_all_request()
            elif choice == "5":
                self.service_request.find_request()
            elif choice == "6":
                self.service_request.change_request_status()
            elif choice == "7":
                pass
            elif choice == "8":
                pass
            elif choice == "9":
                pass
            elif choice == "10":
                pass
            elif choice == "0":
                break

material_service = MaterialService()
request_service = RequestService()
ConsoleMenu(material_service, request_service).main()        


````
