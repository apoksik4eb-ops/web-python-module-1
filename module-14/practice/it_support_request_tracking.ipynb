## Учёт заявок в IT-поддержке”

В компании есть небольшая IT-поддержка. Сотрудники отправляют заявки: `не работает принтер`, `забыл пароль`, `нет интернета`. Нужно сделать консольную программу, которая помогает хранить и обрабатывать эти заявки.

### Что должна уметь программа

Меню:
1. Добавить заявку
2. Показать все заявки
3. Изменить статус заявки
4. Показать открытые заявки
5. Найти заявку по имени сотрудника
0. Выход

### Данные одной заявки
```txt
Номер заявки
Имя сотрудника
Описание проблемы
Статус
```

**Статусы можно сделать такими:**
`новая`, `в работе`, `закрыта`


### Пример заявки
```txt
№1
Сотрудник: Иван
Проблема: Не работает принтер
Статус: новая
```

### Обязательные условия
1. Номер заявки создаётся автоматически.
2. При добавлении заявки статус всегда `новая`.
3. Можно изменить статус только на:
    - `новая`
    - `в работе`
    - `закрыта`
4. При поиске по имени программа должна показывать все заявки этого сотрудника.
5. Открытые заявки — это заявки со статусом `новая` или `в работе`.
6. Решение должно быть реализовано в объектно-ориентированном стиле с соблюдением принципов `SOLID`:
   - отдельный класс для заявки;
   - отдельный класс или сервис для управления заявками;
   - отдельный класс для работы с консольным меню;
   - классы не должны брать на себя лишние обязанности;
   - зависимости между частями программы должны быть минимальными и понятными.

````python
class Ticket:
    def __init__(self, id, name, problem, status="новая"):
        self.ticket_id = id
        self.name = name
        self.problem = problem
        self.status = status

    def __str__(self):
        return (
            f"ID: {self.ticket_id}\n"
            f"Name: {self.name}\n"
            f"Problem: {self.problem}\n"
            f"Status: {self.status}"
        )
    

class TicketService:
    def __init__(self):
        self.tickets_base = []
        self.next_ticket_id = 1


    def add_ticket(self, name, problem):
        ticket = Ticket(self.next_ticket_id, name, problem)
        self.tickets_base.append(ticket)
        print(f"Заявка №{self.next_ticket_id} создана. Сотрудник: {name}, описание проблемы: {problem}")
        self.next_ticket_id += 1      

    def show_all_ticket(self):
        if not self.tickets_base:
            print("Список заявок пуст")
            return
        
        for ticket in self.tickets_base:
            print(ticket)

    
    def change_ticket_status(self):
        ticket_id = int(input("Введите номер заявки: "))

        for ticket in self.tickets_base:
            if ticket.ticket_id == ticket_id:
                print("1. новая\n2. в работе\n3. закрыта")
                choice = input("Выберите статус: ")

                if choice == "1":
                    ticket.status = "новая"
                elif choice == "2":
                    ticket.status = "в работе"
                elif choice == "3":
                    ticket.status = "закрыта"

                print("Статус обновлен")
                return

        print("Заявка не найдена")
        

    def show_open_ticket(self):
        for ticket in self.tickets_base:
            if ticket.status in {"новая", "в работе"}:
                print(ticket)

        print("Список открытых заявок пуст.")


    def find_by_name(self):
        find_ticket = input("Введите имя сотрудника для поиска заявок: ")

        for ticket in self.tickets_base:
            if ticket.name == find_ticket:
                print(ticket)

        print(f"Заявок сотрудника {find_ticket} не найдено.")
        


class ConsoleMenu:
    def __init__(self, ticket_service: TicketService):
        self.service = ticket_service

    def add_ticket(self):
        name = input("Введите имя сотрудника: ")
        problem = input("Введите описание проблемы: ")
        self.service.add_ticket(name, problem)


    def show_menu(self):
        print("1. Добавить заявку")
        print("2. Показать все заявки")
        print("3. Изменить статус заявки")
        print("4. Показать открытые заявки")
        print("5. Найти заявку по имени сотрудника")
        print("6. Выход")

    def main(self):
        while True:
            self.show_menu()

            choice = input("Выберите пункт меню: ")

            if choice == "1":
                self.add_ticket()
            elif choice == "2":
                self.service.show_all_ticket()
            elif choice == "3":
                self.service.change_ticket_status()
            elif choice == "4":
                self.service.show_open_ticket()
            elif choice == "5":
                self.service.find_by_name()
            elif choice == "6":
                break

ticket_service = TicketService()
ConsoleMenu(ticket_service).main()
````
