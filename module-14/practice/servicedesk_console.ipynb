# ЗАДАНИЕ: Консольное приложение для учета заявок на обслуживание оборудования

Необходимо разработать консольное приложение ServiceDesk Console для учета заявок на техническое обслуживание оборудования в организации.

В организации сотрудники используют различное оборудование: ноутбуки, компьютеры, принтеры, роутеры, сканеры и другую технику. Если оборудование выходит из строя или требует обслуживания, сотрудник оставляет заявку. Приложение должно позволять вести учет таких заявок, отслеживать их статус и получать краткую статистику.

## Цель работы:
Создать консольную программу, которая позволяет добавлять, просматривать, искать, редактировать и удалять заявки на обслуживание оборудования, а также сохранять данные между запусками программы.

## Функциональные требования:

### 1. Главное меню

При запуске программы пользователь должен видеть меню:

```txt
=== ServiceDesk Console ===

1. Добавить заявку
2. Показать все заявки
3. Найти заявку
4. Изменить статус заявки
5. Удалить заявку
6. Показать статистику
7. Сохранить данные
0. Выход
```
Пользователь выбирает действие путем ввода номера пункта меню.

### 2. Добавление заявки

Программа должна позволять добавить новую заявку.

При создании заявки пользователь вводит следующие данные:

```txt
- ФИО сотрудника
- отдел сотрудника
- тип оборудования
- инвентарный номер оборудования
- описание проблемы
- приоритет заявки
- статус заявки
```

**Приоритет заявки может принимать одно из значений:**

- низкий
- средний
- высокий

**Статус заявки может принимать одно из значений:**

- новая
- в работе
- выполнена

**Каждой заявке должен автоматически присваиваться уникальный номер.**

**Пример заявки:**
```txt
Номер заявки: 1
Сотрудник: Иванов Иван Иванович
Отдел: Бухгалтерия
Оборудование: Принтер
Инвентарный номер: PR-1023
Описание проблемы: Не печатает документы
Приоритет: высокий
Статус: новая
```

### 3. Просмотр всех заявок

Программа должна выводить список всех созданных заявок.

Для каждой заявки необходимо отображать:
```txt
- номер заявки
- ФИО сотрудника
- отдел
- тип оборудования
- инвентарный номер
- описание проблемы
- приоритет
- статус
```

Если заявок нет, программа должна вывести сообщение:

`Заявки отсутствуют.`

### 4. Поиск заявок

Программа должна позволять искать заявки по разным параметрам.

Необходимо реализовать поиск:
```txt
- по ФИО сотрудника
- по отделу
- по статусу
- по инвентарному номеру оборудования
```

При выборе пункта `Найти заявку` пользователь должен выбрать критерий поиска, затем ввести значение для поиска.

Пример меню поиска:
```txt
1. Поиск по сотруднику
2. Поиск по отделу
3. Поиск по статусу
4. Поиск по инвентарному номеру
```
Если подходящие заявки найдены, программа выводит их список.
Если заявок не найдено, программа выводит сообщение:

`Заявки не найдены.`

### 5. Изменение статуса заявки

Программа должна позволять изменить статус существующей заявки.

Для изменения статуса пользователь вводит номер заявки, после чего выбирает новый статус:
```txt
1. новая
2. в работе
3. выполнена
```

Если заявка с указанным номером существует, программа изменяет ее статус и выводит сообщение об успешном изменении.

Если заявка не найдена, программа выводит сообщение:

`Заявка с указанным номером не найдена.`

### 6. Удаление заявки

Программа должна позволять удалить заявку.

Для удаления пользователь вводит номер заявки.

Если заявка с таким номером существует, программа удаляет ее и выводит сообщение:

`Заявка успешно удалена.`

Если заявка не найдена, программа выводит сообщение:

`Заявка с указанным номером не найдена.`

### 7. Статистика

Программа должна выводить краткую статистику по заявкам:

- общее количество заявок
- количество новых заявок
- количество заявок в работе
- количество выполненных заявок
- количество заявок с высоким приоритетом

Пример вывода:
```txt
=== Статистика ===

Всего заявок: 10
Новые: 3
В работе: 4
Выполнены: 3
С высоким приоритетом: 2
```

### 8. Сохранение данных

Программа должна сохранять данные о заявках в файл.

Формат файла можно выбрать самостоятельно:

- txt
- csv
- json

После сохранения программа должна вывести сообщение:

`Данные успешно сохранены.`

### 9. Загрузка данных

При запуске программы необходимо загружать ранее сохраненные заявки из файла, если такой файл существует.

Если файл отсутствует, программа начинает работу с пустым списком заявок.

### 10. Выход из программы

При выборе пункта `0. Выход` программа должна завершить работу.

Перед выходом желательно выполнить автоматическое сохранение данных в файл.

Требования к обработке ошибок:

1. Программа не должна завершаться аварийно при некорректном вводе пользователя.
2. Если пользователь вводит неправильный пункт меню, необходимо вывести сообщение:

`Некорректный выбор. Попробуйте снова.`

3. Если пользователь вводит несуществующий номер заявки, необходимо вывести сообщение:

`Заявка с указанным номером не найдена.`

4. Если пользователь вводит недопустимый статус или приоритет, программа должна попросить повторить ввод.

Требования к структуре программы:

В программе рекомендуется использовать:

- класс или структуру для хранения данных заявки
- список или коллекцию для хранения всех заявок
- отдельные функции или методы для:
  - вывода меню
  - добавления заявки
  - просмотра заявок
  - поиска заявок
  - изменения статуса
  - удаления заявки
  - вывода статистики
  - сохранения данных
  - загрузки данных

Пример структуры заявки:

Заявка:
- id
- employeeName
- department
- equipmentType
- inventoryNumber
- problemDescription
- priority
- status

Дополнительные требования:

1. Номер заявки должен быть уникальным.
2. После удаления заявки номера остальных заявок не должны изменяться.
3. При добавлении новой заявки номер должен быть больше, чем у всех уже существующих заявок.
4. Поиск должен работать без учета регистра букв.
5. Данные должны сохраняться между запусками программы.

Результат выполнения работы:

В результате должно быть создано консольное приложение ServiceDesk Console, которое позволяет вести учет заявок на обслуживание оборудования в организации и сохранять информацию о заявках между запусками программы.

````python
import json

class Ticket:
    def __init__(self, id, employeeName, department, equipmentType, inventoryNumber, problemDescription, priority, status="новая"):
        self.id = id
        self.employeeName = employeeName
        self.department = department
        self.equipmentType = equipmentType
        self.inventoryNumber = inventoryNumber
        self.problemDescription = problemDescription
        self.priority = priority
        self.status = status

    def __str__(self):
        return (
            f"Номер заявки: {self.id}\n"
            f"Сотрудник: {self.employeeName}\n"
            f"Отдел: {self.department}\n"
            f"Оборудование: {self.equipmentType}\n"
            f"Инвентарный номер: {self.inventoryNumber}\n"
            f"Описание проблемы: {self.problemDescription}\n"
            f"Приоритет: {self.priority}\n"
            f"Статус: {self.status}\n"
        )
    
    def to_dict(self):
        return {
            "id": self.id,
            "employeeName": self.employeeName,
            "department": self.department,
            "equipmentType": self.equipmentType,
            "inventoryNumber": self.inventoryNumber,
            "problemDescription": self.problemDescription,
            "priority": self.priority,
            "status": self.status
        }
    
class TicketService:
    def __init__(self):
        self.tickets = []
        self.next_ticket_id = 1

    def add_ticket(self, employeeName, department, equipmentType, inventoryNumber, problemDescription, priority):
        ticket = Ticket(self.next_ticket_id, employeeName, department, equipmentType, inventoryNumber, problemDescription, priority)
        self.tickets.append(ticket)
        print(f"Заявка {self.next_ticket_id} создана.")
        self.next_ticket_id += 1
        return ticket.id
        

    def show_all_ticket(self):
        if not self.tickets:
            print("Заявки отсутствуют.")
            return
        
        print("Все заявки:\n")
        for ticket in self.tickets:
            print(ticket)
    
    def find_ticket(self):
        choice = input(f"Выберите кретерий поиска:\n1. Поиск по сотруднику\n2. Поиск по отделу\n3. Поиск по статусу\n4. Поиск по инвентарному номеру\n")
        
        found_ticket = []

        if choice == "1":
            find = input("Введите ФИО сотрудника: ")
            for ticket in self.tickets:
                if ticket.employeeName == find:
                    found_ticket.append(ticket)
        elif choice == "2":
            find = input("Введите название отдела: ")
            for ticket in self.tickets:
                if ticket.department == find:
                    found_ticket.append(ticket)
        elif choice == "3":
            find = input("Введите статус заявки: ")
            for ticket in self.tickets:
                if ticket.status == find:
                    found_ticket.append(ticket)
        elif choice == "4":
            find = input("Введите инвентарный номер: ")
            for ticket in self.tickets:
                if ticket.inventoryNumber == find:
                    found_ticket.append(ticket)
        else:
            print("Неправильный выбор")
            return
        
        if found_ticket:
            print("Найденные заявки по вашему запросу:\n")
            for ticket in found_ticket:
                print(ticket)
        else:
            print("Заявки не найдены.")

    def change_ticket_status(self):
        change_ticket = int(input("Введите номер заявки: "))

        for ticket in self.tickets:
            if ticket.id == change_ticket:
                print("1. новая\n2. в работе\n3. выполнена\n")
                choice = input("Выберите статус: ")

                if choice == "1":
                    ticket.status = "новая"
                elif choice == "2":
                    ticket.status = "в работе"
                elif choice == "3":
                    ticket.status = "выполнена"

                print("Статус обновлен")
                return
            
        print("Заявка c указанным номером не найдена.")

    def delete_ticket(self):
        del_ticket = int(input("Введите номер заявки: "))

        for ticket in self.tickets:
            if ticket.id == del_ticket:
                self.tickets.remove(ticket)
                print(f"Заявка {ticket.id} успешно удалена.")
                return
            
        print("Заявка c указанным номером не найдена.")

    def statistic_ticket(self):
        total_ticket = 0
        new_ticket = 0
        work_ticket = 0
        done_ticket = 0
        high_ticket = 0

        for ticket in self.tickets:
            total_ticket += 1
            if ticket.status == "новая":
                new_ticket += 1
            elif ticket.status == "в работе":
                work_ticket += 1
            elif ticket.status == "выполнена":
                done_ticket += 1
            
            if ticket.priority == "высокий":
                high_ticket += 1

        print("=== Статистика ===\n")
        print(f"Всего заявок: {total_ticket}")
        print(f"Новые: {new_ticket}")
        print(f"В работе: {work_ticket}")
        print(f"Выполнены: {done_ticket}")
        print(f"С высоким приоритетом: {high_ticket}")


class FileSave:
    def __init__(self, service):
        self.service = service

    def save_json(self, filename = "tickets.json"):
        data = [ticket.to_dict() for ticket in self.service.tickets]

        with open(filename, "w", encoding="utf-8") as file:
            json.dump(data, file, ensure_ascii=False, indent=4)

    def load_json(self, filename = "tickets.json"):
        try:
            with open(filename, "r", encoding="utf-8") as file:
                data = json.load(file)

            self.service.tickets = [Ticket(item["id"], item["employeeName"], item["department"], item["equipmentType"], item["inventoryNumber"], item["problemDescription"], item["priority"], item["status"]) for item in data]

            if self.service.tickets:
                self.service.next_ticket_id = max(t.id for t in self.service.tickets) + 1

            print("JSON загружен.")

        except FileNotFoundError:
            print("JSON файл не найден.")

    def save_txt(self, filename = "report.txt"):
        with open(filename, "w", encoding="utf-8") as file:
            for ticket in self.service.tickets:
                file.write(str(ticket))
                file.write("\n" + "-" * 50 + "\n")
    
    def save(self):
        self.save_json()
        self.save_txt()
        print("Данные успешно сохранены.")

class ConsoleMenu:
    def __init__(self, ticket_service: TicketService):
        self.service = ticket_service
        self.file_save = FileSave(self.service)
        self.file_save.load_json()

    def add_ticket(self):
        employeeName = input("Введите ФИО сотрудника: ")
        department = input("Введите отдел сотрудника: ")
        equipmentType = input("Введите тип оборудования: ")
        inventoryNumber = input("Введите инвентарный номер оборудования: ")
        problemDescription = input("Введите описание проблемы: ")
        choice_priority = input(f"Введите приоритет заявки:\n1. низкий\n2. средний\n3. высокий")
        if choice_priority == "1":
            priority = "низкий"
        elif choice_priority == "2":
            priority = "средний"
        elif choice_priority == "3":
            priority = "высокий"
        else:
            print("Неверный выбор приоритета")
            return
        self.service.add_ticket(employeeName, department, equipmentType, inventoryNumber, problemDescription, priority)
     
    def show_menu(self):
        print("\n=== ServiceDesk Console ===\n")
        print("1. Добавить заявку")
        print("2. Показать все заявки")
        print("3. Найти заявку")
        print("4. Изменить статус заявки")
        print("5. Удалить заявку")
        print("6. Показать статистику")
        print("7. Сохранить данные")
        print("0. Выход")

    def main(self):
        while True:
            self.show_menu()

            choice = input("Выберите пункт меню: ")

            if choice == "1":
                self.add_ticket()
            elif choice == "2":
                self.service.show_all_ticket()
            elif choice == "3":
                self.service.find_ticket()
            elif choice == "4":
                self.service.change_ticket_status()
            elif choice == "5":
                self.service.delete_ticket()
            elif choice == "6":
                self.service.statistic_ticket()
            elif choice == "7":
                self.file_save.save()
            elif choice == "0":
                self.file_save.save()
                break

ticket_service = TicketService()
ConsoleMenu(ticket_service).main()
````
